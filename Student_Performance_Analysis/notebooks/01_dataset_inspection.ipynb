# 01 — Dataset Inspection

**Project:** Student Attendance and Internal Assessment Performance Analysis  
**Dataset:** Student Performance and Clustering Dataset (Kaggle, Muhammad Khubaib Ahmad)  
**Purpose:** Inspect the raw dataset. No cleaning is done here.

```python
# Cell 1 — Import libraries
import pandas as pd
import numpy as np
import os
```

```python
# Cell 2 — Load the dataset
DATA_PATH = os.path.join('..', 'data', 'original', 'student_performance.csv')

df = pd.read_csv(DATA_PATH)
print('Dataset loaded successfully.')
print('Path used:', DATA_PATH)
```

```python
# Cell 3 — First five rows
df.head()
```

```python
# Cell 4 — Last five rows
df.tail()
```

```python
# Cell 5 — Dataset shape (rows, columns)
rows, cols = df.shape
print(f'Rows    : {rows}')
print(f'Columns : {cols}')
```

```python
# Cell 6 — Column names
print('Column names:')
for i, col in enumerate(df.columns, start=1):
    print(f'  {i:2}. {col}')
```

```python
# Cell 7 — Data types
print('Data types:')
print(df.dtypes)
```

```python
# Cell 8 — Missing values per column
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)

missing_report = pd.DataFrame({
    'Missing Count': missing,
    'Missing %': missing_pct
})

print('Missing values by column:')
print(missing_report[missing_report['Missing Count'] > 0].to_string())

if missing.sum() == 0:
    print('No missing values found (except possibly assignments_submitted).')

print(f'\nTotal missing cells: {missing.sum()}')
```

```python
# Cell 9 — Duplicate records
num_duplicates = df.duplicated().sum()
print(f'Number of duplicate rows: {num_duplicates}')

if num_duplicates > 0:
    print('Duplicate rows:')
    print(df[df.duplicated()])
else:
    print('No duplicate rows found.')
```

```python
# Cell 10 — Descriptive statistics for numeric columns
pd.set_option('display.float_format', '{:.2f}'.format)
df.describe()
```

```python
# Cell 11 — Unique values in categorical columns
categorical_cols = df.select_dtypes(include='object').columns.tolist()
print('Categorical columns:', categorical_cols)

for col in categorical_cols:
    print(f'\n--- {col} ---')
    print(df[col].value_counts())
```

```python
# Cell 12 — Numerical value ranges
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()

range_report = pd.DataFrame({
    'Min': df[numeric_cols].min(),
    'Max': df[numeric_cols].max(),
    'Mean': df[numeric_cols].mean().round(2),
    'Std': df[numeric_cols].std().round(2)
})

print('Numerical column ranges:')
print(range_report.to_string())
```

```python
# Cell 13 — Identify attendance and internal assessment variables

attendance_cols = ['total_lectures', 'lectures_attended', 'total_lab_sessions', 'labs_attended']
internal_assessment_cols = ['quiz1_marks', 'quiz2_marks', 'quiz3_marks', 'midterm_marks']
outcome_cols = ['final_marks', 'previous_gpa']
assignment_cols = ['total_assignments', 'assignments_submitted']

print('=== ATTENDANCE VARIABLES ===')
print(df[attendance_cols].describe().round(2))

print('\n=== INTERNAL ASSESSMENT VARIABLES ===')
print(df[internal_assessment_cols].describe().round(2))

print('\n=== OUTCOME VARIABLES ===')
print(df[outcome_cols].describe().round(2))

print('\n=== ASSIGNMENT COLUMN CHECK ===')
print(f'assignments_submitted — all NaN: {df["assignments_submitted"].isnull().all()}')
print(f'assignments_submitted — missing count: {df["assignments_submitted"].isnull().sum()} out of {len(df)}')
print(f'total_assignments — unique values: {df["total_assignments"].unique()}')

print('\n=== QUICK ATTENDANCE RATE PREVIEW ===')
df['lecture_attendance_rate'] = df['lectures_attended'] / df['total_lectures']
df['lab_attendance_rate'] = df['labs_attended'] / df['total_lab_sessions']
print(df[['lecture_attendance_rate', 'lab_attendance_rate']].describe().round(3))

# Drop the derived columns so the original data stays clean
df.drop(columns=['lecture_attendance_rate', 'lab_attendance_rate'], inplace=True)
print('\nNote: derived columns dropped. Original data unchanged.')
```
